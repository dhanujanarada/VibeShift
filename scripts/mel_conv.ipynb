```python
import torch
import torchaudio
import torchaudio.transforms as T
import yaml
import os
from pathlib import Path

from utills.mel import save_mel_with_metadata 
```

```python
rock_path = Path("../data/rock_instru")
non_rock_path = Path("../data/classical")   
rock_output_path = Path("../data/output/rock_mel")
non_rock_output_path = Path("../data/output/classical_mel")
midi_audio =  Path("../data/batch_output/audio")
midi_audio_mel =  Path("../data/batch_output/audio_mel")
```

# Rock Processing

```python
# Get all audio files from rock directory
audio_extensions = ['.wav', '.flac']
audio_files = [f for f in midi_audio.rglob('*') if f.suffix.lower() in audio_extensions]

print(f"Looking in: {midi_audio.absolute()}")
print(f"Directory exists: {midi_audio.exists()}")
print(f"Found {len(audio_files)} audio files")
```

```python
print(f"Found {len(audio_files)} audio files")
print(f"First 5 files: {[f.name for f in audio_files[:5]]}")
```

```python

# Process each audio file
for audio_file in audio_files:
    print(f"Processing: {audio_file.name}")
    
    # Create output filename
    output_file = midi_audio_mel / f"{audio_file.stem}_mel.pt"
    
    # Save mel spectrogram with metadata
    # genre_id: 1 for rock, 0 for non-rock
    save_mel_with_metadata(
        audio_path=audio_file,
        output_path=output_file,
        genre_id=2,  # 2 for synth
        bpm=None  # Optional: add BPM if available
    )

print(f"Processed {len(audio_files)} rock audio files")
```

```python
audio_nonrock_files = [f for f in non_rock_path.rglob('*') if f.suffix.lower() in audio_extensions]

for audio_file in audio_nonrock_files:
    print(f"Processing: {audio_file.name}")
    
    # Create output filename
    output_file = non_rock_output_path / f"{audio_file.stem}_mel.pt"
    
    # Save mel spectrogram with metadata
    # genre_id: 1 for rock, 0 for non-rock
    save_mel_with_metadata(
        audio_path=audio_file,
        output_path=output_file,
        genre_id= 0,  # 0 for non-rock
        bpm=None  # Optional: add BPM if available
    )

print(f"Processed {len(audio_nonrock_files)} non-rock audio files")
```
