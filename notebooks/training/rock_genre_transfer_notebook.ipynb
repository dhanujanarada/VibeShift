# Rock → Synth → Genre Transfer Pipeline
Complete pipeline: Rock audio → MIDI → Synth → DAC latent → Genre transfer → Audio

```python
# Imports
import sys
import torch
import torch.nn as nn
from pathlib import Path
import librosa
import soundfile as sf
import IPython.display as ipd
from tqdm import tqdm

# Add project root to path
sys.path.insert(0, str(Path.cwd()))

from utills.audio_midi_converter import AudioMidiConverter
from models.flow import FlowMatching
from models.dit import DiT

print("✓ Imports loaded")
print(f"PyTorch version: {torch.__version__}")
print(f"CUDA available: {torch.cuda.is_available()}")
```

```python
# Configuration
# Paths
INPUT_ROCK_AUDIO = r"C:\Users\Dhanuja\Desktop\Vibeshift\rock_instru\00015_instrum.flac"
SOUNDFONT_PATH = "data/TimGM6mb.sf2"
CHECKPOINT_PATH = "checkpoints/checkpoints1/best.pt"
OUTPUT_DIR = "output/rock_genre_transfer"

# Model parameters (must match training)
DIT_CONFIG = {
    "input_dim": 1024,
    "embed_dim": 512,
    "num_blocks": 8,
    "num_heads": 8,
    "num_genres": 3,
    "hidden_dim": 2048
}


# Genre transfer settings
TARGET_GENRE = 1  # 0=synth, 1=classical
NUM_INFERENCE_STEPS = 50
GENRE_NAMES = {0: "synth", 1: "classical"}

# Device
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

# Create output directory
Path(OUTPUT_DIR).mkdir(parents=True, exist_ok=True)

print(f"Device: {DEVICE}")
print(f"Target genre: {GENRE_NAMES[TARGET_GENRE]}")
print(f"Output: {OUTPUT_DIR}")
```

```python
# Step 1: Rock Audio → MIDI → Synth Audio
print(f"\n{'='*70}")
print("STEP 1: Rock → MIDI → Synth")
print(f"{'='*70}")

converter = AudioMidiConverter(soundfont_path=SOUNDFONT_PATH)

# Convert to MIDI
print("\n[1.1] Converting rock audio to MIDI...")
midi_path = converter.audio_to_midi(
    audio_path=INPUT_ROCK_AUDIO,
    output_dir=OUTPUT_DIR
)
print(f"✓ MIDI: {midi_path}")

# Convert MIDI to synth audio
print("\n[1.2] Synthesizing MIDI with soundfont...")
input_name = Path(INPUT_ROCK_AUDIO).stem
synth_audio_path = Path(OUTPUT_DIR) / f"{input_name}_synth.wav"

converter.midi_to_audio(
    midi_path=midi_path,
    output_path=str(synth_audio_path)
)
print(f"✓ Synth audio: {synth_audio_path}")

print("\n🎹 Listen to synth version:")
synth_audio, sr = librosa.load(str(synth_audio_path), sr=None)
ipd.display(ipd.Audio(synth_audio, rate=sr))
```

```python
# Step 2: Synth Audio → DAC Latent
print(f"\n{'='*70}")
print("STEP 2: Synth Audio → DAC Latent Encoding")
print(f"{'='*70}")

# Load DAC encoder
import dac
import torchaudio
from dac.utils import load_model

print("Loading DAC model...")
dac_model = load_model(model_type="44khz")  # Automatically downloads if needed
dac_model.to(DEVICE)
dac_model.eval()
print("✓ DAC loaded")

# Load and preprocess audio
print("\nEncoding synth audio to latent...")
synth_audio_tensor, sr = torchaudio.load(str(synth_audio_path))

# Resample to 44100 Hz if needed
if sr != 44100:
    print(f"Resampling from {sr} Hz to 44100 Hz...")
    resampler = torchaudio.transforms.Resample(orig_freq=sr, new_freq=44100)
    synth_audio_tensor = resampler(synth_audio_tensor)
    sr = 44100

# Convert stereo to mono
if synth_audio_tensor.shape[0] == 2:
    print("Converting stereo to mono...")
    synth_audio_tensor = synth_audio_tensor.mean(dim=0, keepdim=True)  # (1, T)
elif synth_audio_tensor.shape[0] > 2:
    synth_audio_tensor = synth_audio_tensor.mean(dim=0, keepdim=True)

print(f"Audio shape: {synth_audio_tensor.shape} (channels, samples)")
print(f"Sample rate: {sr} Hz")

synth_audio_tensor = synth_audio_tensor.to(DEVICE)

# Encode to latent
with torch.no_grad():
    # DAC expects (batch, channels, samples) - add batch dimension
    if synth_audio_tensor.dim() == 2:  # (C, T)
        synth_audio_tensor = synth_audio_tensor.unsqueeze(0)  # (1, C, T)
    
    # Encode
    dac_output = dac_model.encode(synth_audio_tensor)
    
    # Extract latent codes
    # DAC returns (z, codes, latents, commitment_loss, codebook_loss)
    z = dac_output[0]  # Latent representation
    
    # Reshape to (B, T, D) for model
    if z.dim() == 3:  # (B, D, T)
        source_latent = z.permute(0, 2, 1)  # (B, T, D)
    else:
        source_latent = z

print(f"✓ Latent encoded: {source_latent.shape}")
print(f"  Batch: {source_latent.shape[0]}")
print(f"  Time steps: {source_latent.shape[1]}")
print(f"  Embedding dim: {source_latent.shape[2]}")

# Save synth latent
synth_latent_path = Path(OUTPUT_DIR) / f"{input_name}_synth_latent.pt"
torch.save(source_latent.cpu(), synth_latent_path)
print(f"✓ Saved: {synth_latent_path}")
```

```python
# Step 3: Load Trained Genre Transfer Model
print(f"\n{'='*70}")
print("STEP 3: Load Genre Transfer Model")
print(f"{'='*70}")

print("Loading checkpoint...")
checkpoint = torch.load(CHECKPOINT_PATH, map_location=DEVICE)

# Check checkpoint structure
if 'model_state_dict' in checkpoint:
    state_dict = checkpoint['model_state_dict']
    epoch = checkpoint.get('epoch', 'unknown')
    loss = checkpoint.get('loss', 'unknown')
    print(f"  Checkpoint epoch: {epoch}")
    print(f"  Training loss: {loss}")
else:
    state_dict = checkpoint

# Create model
dit = DiT(**DIT_CONFIG)

flow_model = FlowMatching(dit)
flow_model.load_state_dict(state_dict)
flow_model.to(DEVICE)
flow_model.eval()

print("✓ Model loaded")
print(f"  Parameters: {sum(p.numel() for p in flow_model.parameters()) / 1e6:.2f}M")
```

```python
# Step 4: Genre Transfer (Synth → Target Genre)
print(f"\n{'='*70}")
print(f"STEP 4: Genre Transfer (Synth → {GENRE_NAMES[TARGET_GENRE].upper()})")
print(f"{'='*70}")

@torch.no_grad()
def transfer_genre(
    source_latent: torch.Tensor,
    target_genre_id: int,
    num_steps: int = 50
) -> torch.Tensor:
    """
    Transfer latent to target genre using flow matching.
    
    Args:
        source_latent: (B, T, D) source latent
        target_genre_id: Target genre (0=synth, 1=classical)
        num_steps: Number of flow matching steps
    
    Returns:
        transferred_latent: (B, T, D) transferred latent
    """
    B, T, D = source_latent.shape
    
    # Genre tensor
    genre_ids = torch.full((B,), target_genre_id, dtype=torch.long, device=DEVICE)
    
    # Flow matching: interpolate from source (t=0) to target (t=1)
    dt = 1.0 / num_steps
    x = source_latent.clone()
    
    for step in tqdm(range(num_steps), desc="Genre transfer"):
        t = torch.full((B,), step * dt, device=DEVICE)
        
        # Predict velocity field
        v_pred = flow_model.dit(x, t, genre_ids)
        
        # Update along flow
        x = x + v_pred * dt
    
    return x

# Run transfer
print(f"Transferring to {GENRE_NAMES[TARGET_GENRE]} style...")
transferred_latent = transfer_genre(
    source_latent=source_latent,
    target_genre_id=TARGET_GENRE,
    num_steps=NUM_INFERENCE_STEPS
)

print(f"\n✓ Transfer complete: {transferred_latent.shape}")

# Save transferred latent
transferred_latent_path = Path(OUTPUT_DIR) / f"{input_name}_to_{GENRE_NAMES[TARGET_GENRE]}_latent.pt"
torch.save(transferred_latent.cpu(), transferred_latent_path)
print(f"✓ Saved: {transferred_latent_path}")
```

```python
# Step 5: DAC Latent → Audio (Decode)
print(f"\n{'='*70}")
print("STEP 5: DAC Latent → Audio Decoding")
print(f"{'='*70}")

print("Decoding transferred latent to audio...")

# Reshape back to (B, D, T) for DAC decoder
if transferred_latent.dim() == 3:  # (B, T, D)
    z_for_decode = transferred_latent.permute(0, 2, 1)  # (B, D, T)
else:
    z_for_decode = transferred_latent

with torch.no_grad():
    # Decode
    decoded_audio = dac_model.decode(z_for_decode)
    
    # decoded_audio is (B, C, T) - convert to numpy
    if decoded_audio.dim() == 3:
        decoded_audio = decoded_audio.squeeze(0)  # Remove batch
    if decoded_audio.dim() == 2:
        decoded_audio = decoded_audio.mean(dim=0)  # Mono if stereo
    
    audio_np = decoded_audio.cpu().numpy()

# Save audio
output_audio_path = Path(OUTPUT_DIR) / f"{input_name}_to_{GENRE_NAMES[TARGET_GENRE]}.wav"
sf.write(str(output_audio_path), audio_np, dac_model.sample_rate)

print(f"✓ Decoded audio saved: {output_audio_path}")
print(f"  Sample rate: {dac_model.sample_rate} Hz")
print(f"  Duration: {len(audio_np) / dac_model.sample_rate:.2f}s")
```

```python
# Step 6: Listen & Compare
print(f"\n{'='*70}")
print("STEP 6: Results Playback")
print(f"{'='*70}")

# Original rock
print("\n🎸 Original rock audio:")
orig_audio, sr_orig = librosa.load(INPUT_ROCK_AUDIO, sr=None)
ipd.display(ipd.Audio(orig_audio, rate=sr_orig))

# Synth version
print("\n🎹 Synthesized (from MIDI):")
synth_audio, sr_synth = librosa.load(str(synth_audio_path), sr=None)
ipd.display(ipd.Audio(synth_audio, rate=sr_synth))

# Transferred version
print(f"\n🎻 Transferred to {GENRE_NAMES[TARGET_GENRE]} style:")
transferred_audio, sr_transfer = librosa.load(str(output_audio_path), sr=None)
ipd.display(ipd.Audio(transferred_audio, rate=sr_transfer))

print(f"\n✓ Pipeline complete!")
```

```python
# Summary
print(f"\n{'='*70}")
print("PIPELINE SUMMARY")
print(f"{'='*70}")
print(f"\nInput:  {INPUT_ROCK_AUDIO}")
print(f"\nProcessing steps:")
print(f"  1. Rock audio → MIDI: {midi_path}")
print(f"  2. MIDI → Synth: {synth_audio_path}")
print(f"  3. Synth → Latent: {synth_latent_path}")
print(f"  4. Genre transfer: synth → {GENRE_NAMES[TARGET_GENRE]}")
print(f"  5. Latent → Audio: {output_audio_path}")
print(f"\nOutput directory: {OUTPUT_DIR}")

# List all generated files
output_files = sorted(Path(OUTPUT_DIR).glob("*"))
print(f"\nGenerated {len(output_files)} files:")
for f in output_files:
    size_mb = f.stat().st_size / (1024 * 1024)
    print(f"  - {f.name:50s} ({size_mb:6.2f} MB)")

print(f"\n{'='*70}")
```
